## 3. Post-Execution Validation

```python
print("\n=== Final Validation ===")

kpi_tables = [
    "workspace.gold.kpi_recovery_opportunity_score",
    "workspace.gold.kpi_price_elasticity_impact",
    "workspace.gold.kpi_execution_roi",
    "workspace.gold.kpi_channel_chain_performance",
    "workspace.gold.kpi_product_portfolio_health",
    "workspace.gold.kpi_competitive_position_velocity",
    "workspace.gold.kpi_perfect_storm_index"
]

validation_results = []

for table in kpi_tables:
    try:
        count = spark.sql(f"SELECT COUNT(*) as cnt FROM {table}").collect()[0]['cnt']
        latest_date = spark.sql(f"SELECT MAX(date) as max_date FROM {table}").collect()[0]['max_date']
        validation_results.append({
            'table': table,
            'status': '✓',
            'row_count': count,
            'latest_date': latest_date
        })
        print(f"✓ {table}: {count:,} rows, latest date: {latest_date}")
    except Exception as e:
        validation_results.append({
            'table': table,
            'status': '✗',
            'error': str(e)
        })
        print(f"✗ {table}: ERROR - {str(e)}")

print(f"\n=== Orchestration completed at: {datetime.now()} ===")
```

## 4. Quick Insights Summary

```python
print("\n=== Quick Insights Summary ===")

# Perfect Storms count
perfect_storms = spark.sql("""
    SELECT risk_tier, COUNT(*) as count
    FROM workspace.gold.kpi_perfect_storm_index
    WHERE date = (SELECT MAX(date) FROM workspace.gold.kpi_perfect_storm_index)
    GROUP BY risk_tier
    ORDER BY count DESC
""").toPandas()

print("\nPerfect Storm Distribution:")
print(perfect_storms.to_string(index=False))

# Top recovery opportunities
top_recovery = spark.sql("""
    SELECT recovery_action_type, COUNT(*) as count, SUM(potential_revenue_lost_usd) as total_revenue_at_risk
    FROM workspace.gold.kpi_recovery_opportunity_score
    WHERE date = (SELECT MAX(date) FROM workspace.gold.kpi_recovery_opportunity_score)
    GROUP BY recovery_action_type
    ORDER BY total_revenue_at_risk DESC
""").toPandas()

print("\nTop Recovery Action Types:")
print(top_recovery.to_string(index=False))
```
