```python
from datetime import datetime

# ----------------------------------------
# Logging helper
# ----------------------------------------
def log_step(step_name, status, msg=None):
    timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    print(f"[{timestamp}] [{status}] {step_name}{' - ' + msg if msg else ''}")

# ----------------------------------------
# Run SQL notebook / script safely
# ----------------------------------------
def run_sql(path, step_name):
    try:
        log_step(step_name, "START")
        dbutils.notebook.run(path, 0)
        log_step(step_name, "SUCCESS")
    except Exception as e:
        log_step(step_name, "FAIL", str(e))
        raise e
```
